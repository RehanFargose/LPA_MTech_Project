```python

# Load your enriched dataset
# df = pd.read_parquet(f"D:\LPA_MTech_Project\Enriched_Datasets\SupremeCourt_Combined_{start}_{end}_enriched.parquet")
df = pd.read_parquet(f"D:\LPA_MTech_Project\Enriched_Datasets\SC_{start}_{end}_enriched_IPC.parquet")

# D:\LPA_MTech_Project\Enriched_Datasets\SC_{start}_{end}_enriched_IPC_{dataset_num}.parquet
print("Loaded:", df.shape)
print("\nColumns:", df.columns.tolist())
```

```python
print("\nHead:")
display(df.head(3))
```

```python

# ---------------------
# 2. Missing Data
# ---------------------
print("\n=== TOP MISSING VALUES ===")
print(df.isna().sum().sort_values(ascending=False).head(20))


```

```python
critical_cols = ["case_id", "text", "verdict_label", "decision_date"]
print("\nMissing in critical columns:")
print(df[critical_cols].isna().sum())
```

```python

# ---------------------
# 3. Duplicate Check
# ---------------------
print("\n=== DUPLICATE CHECK ===")

# Case ID is enough in most datasets
print("Duplicate case_id:", df["case_id"].duplicated().sum())

# Safe duplicate check only on hashable columns
hashable_cols = [
    c for c in df.columns
    if df[c].apply(lambda x: isinstance(x, (int, float, str, bool, type(None)))).all()
]

print("Hashable columns used:", hashable_cols)
print("Duplicate rows (hashable only):", df.duplicated(subset=hashable_cols).sum())

```

```python
# ---------------------
# 4. Verdict distribution
# ---------------------
print("\n=== VERDICT LABEL DISTRIBUTION ===")
print(df["verdict_label"].value_counts(normalize=True))
```

```python
# ---------------------
# 5. Case type distribution
# ---------------------
print("\n=== CASE TYPE DISTRIBUTION ===")
print(df["case_type"].value_counts().head(20))
```

```python
# ---------------------
# 6. Text Quality Checks
# ---------------------
print("\n=== TEXT QUALITY ===")
df["text_len"] = df["text"].astype(str).apply(len)
df["word_count"] = df["text"].astype(str).apply(lambda x: len(x.split()))
```

```python

print("Average chars:", df["text_len"].mean())
print("Average words:", df["word_count"].mean())
print("Zero or tiny texts:", (df["word_count"] < 20).sum())
```

```python
# Check HTML artifacts
sample_html = df[df["text"].str.contains("<p>|&nbsp;", regex=True, na=False)].head(3)
print("\nPossible HTML artifacts in text:")
display(sample_html)
```

```python
# ---------------------
# 7. Sections Quality
# ---------------------
sections = ["facts_section", "issues_section", "arguments_section",
            "reasoning_section", "conclusion_section"]

print("\n=== SECTION COMPLETENESS ===")
for sec in sections:
    missing = df[sec].isna().mean() * 100
    print(f"{sec}: {missing:.2f}% missing")

print("\nPreview extracted sections:")
display(df[sections].head(3))
# ['case_id', 'year', 'decision_date', 'case_no', 'bench_info', 'bench_size', 'judges', 'petitioner', 'respondent', 'disposal_nature', 'verdict_label', 'raw_html', 'text', 'decade', 'num_words_text', 'num_lines', 'petitioner_type', 'respondent_type', 'avg_sentence_length', 'num_paragraphs', 'proceeding_type', 'act_names', 'ipc_mentions', 'ipc_mentions_extra', 'ipc_sections', 'ipc_titles', 'num_ipc_refs', 'has_ipc', 'num_unique_acts', 'facts_section', 'issues_section', 'arguments_section', 'reasoning_section', 'conclusion_section', 'sentiment_score', 'petitioner_mentions', 'respondent_mentions', 'pet_vs_resp_ratio', 'keywords', 'case_type', 'citations', 'num_citations', 'complexity_score', 'constitution_articles', 'has_constitution_article']
```

```python
# ---------------------
# 8. IPC Mapping Checks
# ---------------------
print("\n=== IPC FIELD CHECKS ===")
print("has_ipc counts:", df["has_ipc"].value_counts())
```

```python
# inconsistent rows
incon = df[(df["has_ipc"] == False) & (df["num_ipc_refs"] > 0)]
print("Inconsistent IPC flags:", len(incon))
```

```python

print("\nMost common IPC sections:")
all_ipc = []
for lst in df["ipc_sections"].dropna():
    if isinstance(lst, list):
        all_ipc.extend(lst)
print(Counter(all_ipc).most_common(20))
```

```python
# ---------------------
# 9. Numerical Field Quality
# ---------------------
num_cols = ["bench_size", "num_words_text", "avg_sentence_length",
            "num_paragraphs", "num_citations", "complexity_score"]

print("\n=== NUMERIC COLUMN SUMMARY ===")
display(df[num_cols].describe())
```

```python
# detect negative values
for col in num_cols:
    neg = (df[col] < 0).sum()
    if neg > 0:
        print(f"WARNING: {col} has {neg} negative values")
```

```python
# ---------------------
# 10. Keyword & Citation Analysis
# ---------------------
print("\n=== TOP KEYWORDS ===")
try:
    keywords = Counter([kw for lst in df["keywords"].dropna() for kw in lst])
    print(keywords.most_common(30))
except:
    print("Keyword column not list-like")
```

```python
print("\n=== CITATIONS SAMPLE ===")
display(df[["citations", "num_citations"]].head(5))
```

```python
# ---------------------
# 11. Constitution Articles
# ---------------------
print("\n=== CONSTITUTION ARTICLE USAGE ===")
all_const = []
for lst in df["constitution_articles"].dropna():
    if isinstance(lst, list):
        all_const.extend(lst)
print(Counter(all_const).most_common(15))

print("\n=== QUALITY CHECK COMPLETE ===")
```
